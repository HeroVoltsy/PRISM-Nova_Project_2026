```python
chi2, p_value, dof, expected = chi2_contingency(site_table)

n = site_table.to_numpy().sum()
min_dim = min(site_table.shape) - 1
cramers_v = (chi2 / (n * min_dim)) ** 0.5

print(f"Chi-square statistic: {chi2:.4f}")
print(f"Degrees of freedom: {dof}")
print(f"P-value: {p_value:.6g}")
print(f"Cramér's V: {cramers_v:.4f}")
print("Minimum expected cell count:", expected.min())
```

```python
#Site based outliers

X = np.load(x_path, mmap_mode="r")

X_array = np.asarray(X, dtype=np.float32)

group_mean = X_array.mean(axis=0)

# Participant-level RMS deviation from the cohort mean
participant_rmsd = np.sqrt(
    np.mean((X_array - group_mean) ** 2, axis=1)
)

participant_q1, participant_q3 = np.percentile(
    participant_rmsd, [25, 75]
)
participant_iqr = participant_q3 - participant_q1

participant_upper = participant_q3 + 3 * participant_iqr

outlier_mask = participant_rmsd > participant_upper



participant_qc = metadata.copy()
participant_qc["CONNECTIVITY_RMSD"] = participant_rmsd
participant_qc["POTENTIAL_OUTLIER"] = outlier_mask
outliers_by_site = (
    participant_qc.groupby("SITE_ID")
    .agg(
        participants=("POTENTIAL_OUTLIER", "size"),
        outliers=("POTENTIAL_OUTLIER", "sum"),
        median_rmsd=("CONNECTIVITY_RMSD", "median"),
        mean_rmsd=("CONNECTIVITY_RMSD", "mean"),
    )
)

outliers_by_site["outlier_percent"] = (
    100 * outliers_by_site["outliers"]
    / outliers_by_site["participants"]
)

display(
    outliers_by_site.sort_values(
        "outlier_percent", ascending=False
    ).round(3)
)
```

```python
outliers = (
    participant_qc.loc[
        participant_qc["POTENTIAL_OUTLIER"]
    ]
    .sort_values("CONNECTIVITY_RMSD", ascending=False)
    .copy()
)

columns = [
    col for col in [
        "FILE_ID",
        "SITE_ID",
        "DX_GROUP",
        "CONNECTIVITY_RMSD",
        "POTENTIAL_OUTLIER",
    ]
    if col in outliers.columns
]

display(outliers[columns].round(4))

#outliers[columns].to_csv(
    #"08_flagged_participants.csv",
    #index=False
#)
```

```python
qc_analysis = metadata.copy()

group_labels = {1: "Autism", 2: "Control"}

group_vars = [
    c for c in [
        "N_TIMEPOINTS",
        "CONNECTIVITY_RMSD",
        "SITE_RELATIVE_RMSD",
    ]
    if c in qc_analysis.columns
]

group_results = []

for variable in group_vars:
    autism = qc_analysis.loc[
        qc_analysis["DX_GROUP"] == 1, variable
    ].dropna()

    control = qc_analysis.loc[
        qc_analysis["DX_GROUP"] == 2, variable
    ].dropna()

    if len(autism) == 0 or len(control) == 0:
        continue

    u_stat, p_value = mannwhitneyu(
        autism, control, alternative="two-sided"
    )

    # Rank-biserial effect size; positive means autism values tend to be higher.
    rank_biserial = 2 * u_stat / (len(autism) * len(control)) - 1

    group_results.append({
        "variable": variable,
        "autism_n": len(autism),
        "autism_median": autism.median(),
        "control_n": len(control),
        "control_median": control.median(),
        "U_statistic": u_stat,
        "rank_biserial_r": rank_biserial,
        "p_value": p_value,
    })

group_results = pd.DataFrame(group_results)

if not group_results.empty:
    group_results["p_adjusted"] = multipletests(
        group_results["p_value"], method="holm"
    )[1]
    display(group_results.round(5))

# Distribution plots
for variable in group_vars:
    fig, ax = plt.subplots(figsize=(8, 5))
    sns.boxplot(
        data=qc_analysis,
        x="DX_GROUP",
        y=variable,
        hue="DX_GROUP",
        legend=False,
        ax=ax,
    )
    ax.set_xticks([0, 1])
    ax.set_xticklabels(["Autism (1)", "Control (2)"])
    ax.set_title(f"{variable}: Autism vs Control")
    plt.tight_layout()
    plt.show()
```

```python
rng = np.random.default_rng(42)
n_sample_features = min(100, X.shape[1])

sample_feature_idx = np.sort(
    rng.choice(X.shape[1], size=n_sample_features, replace=False)
)

X_sample = np.asarray(X[:, sample_feature_idx], dtype=np.float64)

# Feature correlation matrix
feature_corr = np.corrcoef(X_sample, rowvar=False)

fig, ax = plt.subplots(figsize=(12, 10))
sns.heatmap(
    feature_corr,
    cmap="vlag",
    center=0,
    vmin=-1,
    vmax=1,
    xticklabels=False,
    yticklabels=False,
    ax=ax,
)
ax.set_title(f"Connectivity Correlation Matrix ({n_sample_features} sampled features)")
plt.tight_layout()
plt.show()

# Summarize pairwise correlations, excluding the diagonal.
upper = feature_corr[np.triu_indices_from(feature_corr, k=1)]
upper = upper[np.isfinite(upper)]

print("Sampled feature pairs:", len(upper))
print("Median absolute correlation:", np.median(np.abs(upper)))
print("90th percentile absolute correlation:", np.percentile(np.abs(upper), 90))
print("99th percentile absolute correlation:", np.percentile(np.abs(upper), 99))
print("Fraction with |r| > 0.80:", np.mean(np.abs(upper) > 0.80))
print("Fraction with |r| > 0.95:", np.mean(np.abs(upper) > 0.95))
```

```python
#Variance Inflation Factor
vif_feature_idx = sample_feature_idx[:10]
X_vif = np.asarray(X[:, vif_feature_idx], dtype=np.float64)

# Standardize features and remove any constant columns.
X_vif = StandardScaler().fit_transform(X_vif)
nonconstant = X_vif.std(axis=0) > 1e-10
X_vif = X_vif[:, nonconstant]

vif_rows = []

if X_vif.shape[1] >= 2:
    for j, original_idx in enumerate(vif_feature_idx[nonconstant]):
        vif_rows.append({
            "feature_index": int(original_idx),
            "VIF": variance_inflation_factor(X_vif, j),
        })

    vif_table = pd.DataFrame(vif_rows).sort_values(
        "VIF", ascending=False
    )
    display(vif_table.round(3))
else:
    print("Not enough nonconstant features to calculate VIF.")
```

```python
autism = participant_qc.loc[
    participant_qc["DX_GROUP"] == 1, "CONNECTIVITY_RMSD"
].dropna().to_numpy()

control = participant_qc.loc[
    participant_qc["DX_GROUP"] == 2, "CONNECTIVITY_RMSD"
].dropna().to_numpy()

# Difference in means and pooled-standard-deviation effect size
mean_diff = autism.mean() - control.mean()

pooled_sd = np.sqrt(
    ((len(autism) - 1) * autism.var(ddof=1)
     + (len(control) - 1) * control.var(ddof=1))
    / (len(autism) + len(control) - 2)
)

cohens_d = mean_diff / pooled_sd if pooled_sd > 0 else np.nan

# Statistical test
u_stat, p_value = mannwhitneyu(
    autism, control, alternative="two-sided"
)

# Approximate confidence interval for the mean difference using bootstrap
rng = np.random.default_rng(42)
n_bootstrap = 5000

boot_diffs = np.empty(n_bootstrap)

for i in range(n_bootstrap):
    a = rng.choice(autism, size=len(autism), replace=True)
    c = rng.choice(control, size=len(control), replace=True)
    boot_diffs[i] = a.mean() - c.mean()

ci_low, ci_high = np.percentile(boot_diffs, [2.5, 97.5])

# Volcano-style plot: one comparison, so there is one point
plot_df = pd.DataFrame({
    "effect_size": [cohens_d],
    "p_value": [p_value],
    "comparison": ["Autism vs Control"],
})

plot_df["neg_log10_p"] = -np.log10(
    np.maximum(plot_df["p_value"], np.finfo(float).tiny)
)

fig, ax = plt.subplots(figsize=(9, 6))
ax.scatter(
    plot_df["effect_size"],
    plot_df["neg_log10_p"],
    s=150,
    alpha=0.85
)
ax.annotate(
    "Autism vs Control",
    (cohens_d, -np.log10(max(p_value, np.finfo(float).tiny))),
    xytext=(8, 8),
    textcoords="offset points"
)
ax.axhline(-np.log10(0.05), linestyle="--", linewidth=1)
ax.axvline(0, linestyle=":", linewidth=1)
ax.set_xlabel("Cohen's d (Autism − Control)")
ax.set_ylabel("−log10(p-value)")
ax.set_title("Volcano-Style Plot: Connectivity RMSD by Diagnosis")
plt.tight_layout()
plt.show()

print(f"Autism n={len(autism)}, mean={autism.mean():.4f}, SD={autism.std(ddof=1):.4f}")
print(f"Control n={len(control)}, mean={control.mean():.4f}, SD={control.std(ddof=1):.4f}")
print(f"Cohen's d: {cohens_d:.4f}")
print(f"Mean difference: {mean_diff:.4f}")
print(f"Bootstrap 95% CI for mean difference: [{ci_low:.4f}, {ci_high:.4f}]")
print(f"Mann–Whitney U p-value: {p_value:.6g}")
```
