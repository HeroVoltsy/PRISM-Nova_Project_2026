```python
ROOT = Path.home() / "Projects" / "NOVAProject"
OUT = ROOT / "outputs" / "model_experiments"
RESULTS = OUT / "statistical_analysis"
RESULTS.mkdir(parents=True, exist_ok=True)

metrics = [
    "Accuracy", "BalancedAccuracy", "F1", "MCC",
    "ROC_AUC", "PR_AUC", "Sensitivity", "Specificity"
]

cv = pd.read_csv(OUT / "all_fold_metrics.csv")
loso_sites = pd.read_csv(OUT / "loso_results" / "all_site_metrics.csv")
loso_folds = pd.read_csv(OUT / "loso_results" / "all_fold_metrics.csv")

# 1. Cross-validation descriptive statistics
cv_summary = (
    cv.groupby("Model")[metrics]
    .agg(["mean", "std", "median"])
    .round(4)
)
cv_summary.to_csv(RESULTS / "cv_descriptive_summary.csv")

# 2. LOSO descriptive statistics across held-out sites
loso_summary = (
    loso_folds.groupby("Model")[metrics]
    .agg(["mean", "std", "median", "min", "max"])
    .round(4)
)
loso_summary.to_csv(RESULTS / "loso_descriptive_summary.csv")

# 3. LOSO site-level sample sizes and performance
loso_sites.to_csv(RESULTS / "loso_site_metrics_audited.csv", index=False)

# 4. Check whether the two LOSO metric files agree
comparison = loso_sites.merge(
    loso_folds,
    left_on=["Model", "Site"],
    right_on=["Model", "HeldOutSite"],
    suffixes=("_site_file", "_fold_file"),
    how="outer",
    indicator=True
)

comparison["Accuracy_Difference"] = (
    comparison["Accuracy_site_file"] - comparison["Accuracy_fold_file"]
)

comparison["ROC_AUC_Difference"] = (
    comparison["ROC_AUC_site_file"] - comparison["ROC_AUC_fold_file"]
)

comparison.to_csv(
    RESULTS / "loso_file_consistency_check.csv",
    index=False
)

# 5. Exploratory Friedman tests across models on matched CV folds
friedman_results = []

for metric in metrics:
    pivot = cv.pivot(index="Fold", columns="Model", values=metric).dropna()

    if pivot.shape[0] >= 3 and pivot.shape[1] >= 3:
        statistic, p_value = friedmanchisquare(
            *[pivot[col].values for col in pivot.columns]
        )

        friedman_results.append({
            "Metric": metric,
            "N_Folds": pivot.shape[0],
            "N_Models": pivot.shape[1],
            "Friedman_Statistic": statistic,
            "P_Value": p_value
        })

friedman_df = pd.DataFrame(friedman_results)

if not friedman_df.empty:
    friedman_df["P_Holm"] = multipletests(
        friedman_df["P_Value"], method="holm"
    )[1]

friedman_df.to_csv(
    RESULTS / "exploratory_friedman_tests.csv",
    index=False
)

# 6. Exploratory paired Wilcoxon comparisons between models
pairwise_results = []

for metric in metrics:
    pivot = cv.pivot(index="Fold", columns="Model", values=metric).dropna()

    models = pivot.columns.tolist()

    for i, model_a in enumerate(models):
        for model_b in models[i + 1:]:
            a = pivot[model_a].values
            b = pivot[model_b].values

            if np.allclose(a, b):
                statistic, p_value = 0.0, 1.0
            else:
                statistic, p_value = wilcoxon(
                    a, b, alternative="two-sided"
                )

            pairwise_results.append({
                "Metric": metric,
                "Model_A": model_a,
                "Model_B": model_b,
                "Mean_Difference_A_minus_B": np.mean(a - b),
                "Median_Difference_A_minus_B": np.median(a - b),
                "Wilcoxon_Statistic": statistic,
                "P_Value": p_value,
                "N_Paired_Folds": len(a)
            })

pairwise_df = pd.DataFrame(pairwise_results)

if not pairwise_df.empty:
    pairwise_df["P_Holm"] = multipletests(
        pairwise_df["P_Value"], method="holm"
    )[1]

pairwise_df.to_csv(
    RESULTS / "exploratory_pairwise_wilcoxon.csv",
    index=False
)

# 7. Basic dataset and fold audit
audit = {
    "cv_rows": len(cv),
    "cv_models": cv["Model"].nunique(),
    "cv_folds": cv["Fold"].nunique(),
    "cv_prediction_rows": len(pd.read_csv(OUT / "all_predictions.csv")),
    "loso_site_metric_rows": len(loso_sites),
    "loso_fold_metric_rows": len(loso_folds),
    "loso_models": loso_folds["Model"].nunique(),
    "loso_heldout_sites": loso_folds["HeldOutSite"].nunique(),
    "loso_prediction_rows": len(pd.read_csv(
        OUT / "loso_results" / "all_predictions.csv"
    ))
}

pd.Series(audit).to_csv(RESULTS / "experiment_audit.csv", header=["Value"])

print("\nCV SUMMARY")
print(cv_summary)

print("\nLOSO SUMMARY")
print(loso_summary)

print("\nLOSO FILE CONSISTENCY")
print(comparison[
    ["Model", "Site", "Accuracy_Difference", "ROC_AUC_Difference", "_merge"]
].head(20))

print("\nEXPLORATORY FRIEDMAN TESTS")
print(friedman_df)

print(f"\nResults saved to: {RESULTS}")
```

```python
ROOT = Path.home() / "Projects" / "NOVAProject"
OUT = ROOT / "outputs" / "model_experiments"
LOSO = OUT / "loso_results"
RESULTS = OUT / "statistical_analysis"

RESULTS.mkdir(parents=True, exist_ok=True)

PREDICTION_FILE = LOSO / "all_predictions.csv"

N_BOOTSTRAP = 5000
RANDOM_STATE = 42
QUANTILES = [0.25, 0.50, 0.75]

METRICS = [
    "Accuracy",
    "BalancedAccuracy",
    "Precision",
    "Sensitivity",
    "Specificity",
    "F1",
    "MCC",
    "ROC_AUC",
    "PR_AUC"
]


# ---------------------------------------------------------
# 1. Load and validate predictions
# ---------------------------------------------------------

pred = pd.read_csv(PREDICTION_FILE)

required = {
    "Model", "Fold", "FILE_ID", "HeldOutSite",
    "y_true", "y_pred", "y_score"
}

missing = required - set(pred.columns)

if missing:
    raise ValueError(f"Missing prediction columns: {missing}")

if pred["FILE_ID"].isna().any():
    raise ValueError("Missing FILE_ID values.")

if pred.duplicated(["Model", "FILE_ID"]).any():
    raise ValueError(
        "Duplicate Model/FILE_ID predictions detected. "
        "Resolve repeated predictions before paired testing."
    )

models = sorted(pred["Model"].unique())

if len(models) < 2:
    raise ValueError("At least two models are required.")

print("Models:", models)
print("Prediction rows:", len(pred))
print("Unique participants:", pred["FILE_ID"].nunique())
print("Held-out sites:", pred["HeldOutSite"].nunique())


# ---------------------------------------------------------
# 2. Align models on identical participants
# ---------------------------------------------------------

first_model = models[0]

base = (
    pred[pred["Model"] == first_model]
    [["FILE_ID", "HeldOutSite", "y_true"]]
    .copy()
)

aligned = base.copy()

for model in models:
    current = (
        pred[pred["Model"] == model]
        [["FILE_ID", "HeldOutSite", "y_true", "y_pred", "y_score"]]
        .copy()
    )

    if current["FILE_ID"].nunique() != len(base):
        raise ValueError(
            f"{model} does not have the same participant count "
            "as the reference model."
        )

    check = base.merge(
        current[["FILE_ID", "HeldOutSite", "y_true"]],
        on="FILE_ID",
        suffixes=("_base", "_model"),
        how="outer",
        indicator=True
    )

    if not (check["_merge"] == "both").all():
        raise ValueError(f"Participant mismatch in {model}.")

    if not (
        check["HeldOutSite_base"] == check["HeldOutSite_model"]
    ).all():
        raise ValueError(f"Site mismatch in {model}.")

    if not (check["y_true_base"] == check["y_true_model"]).all():
        raise ValueError(f"True-label mismatch in {model}.")

    current = current.drop(columns=["HeldOutSite", "y_true"])
    current = current.rename(columns={
        "y_pred": f"{model}__pred",
        "y_score": f"{model}__score"
    })

    aligned = aligned.merge(
        current,
        on="FILE_ID",
        how="inner",
        validate="one_to_one"
    )

aligned = aligned.sort_values("FILE_ID").reset_index(drop=True)

print("Aligned participants:", len(aligned))


# ---------------------------------------------------------
# 3. Recompute LOSO metrics from predictions
# ---------------------------------------------------------

site_rows = []

for (model, site), group in pred.groupby(
    ["Model", "HeldOutSite"]
):
    y = group["y_true"].to_numpy()
    yp = group["y_pred"].to_numpy()
    ys = group["y_score"].to_numpy()

    tn, fp, fn, tp = confusion_matrix(
        y, yp, labels=[0, 1]
    ).ravel()

    site_rows.append({
        "Model": model,
        "Site": site,
        "N_Test": len(group),
        "ASD_Prevalence": np.mean(y == 1),
        "Accuracy": accuracy_score(y, yp),
        "BalancedAccuracy": balanced_accuracy_score(y, yp),
        "Precision": precision_score(y, yp, zero_division=0),
        "Sensitivity": recall_score(y, yp, zero_division=0),
        "Specificity": tn / (tn + fp) if tn + fp else np.nan,
        "F1": f1_score(y, yp, zero_division=0),
        "MCC": matthews_corrcoef(y, yp),
        "ROC_AUC": (
            roc_auc_score(y, ys)
            if len(np.unique(y)) == 2 else np.nan
        ),
        "PR_AUC": average_precision_score(y, ys),
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    })

site_metrics = pd.DataFrame(site_rows)

site_metrics.to_csv(
    RESULTS / "10_recomputed_loso_site_metrics.csv",
    index=False
)

site_summary = (
    site_metrics.groupby("Model")[METRICS]
    .agg(["mean", "std", "median", "min", "max"])
)

site_summary.to_csv(
    RESULTS / "10_recomputed_loso_summary.csv"
)


# ---------------------------------------------------------
# 4. DeLong test implementation
# ---------------------------------------------------------

def compute_midrank(x):
    order = np.argsort(x)
    sorted_x = x[order]
    n = len(x)
    midranks = np.zeros(n, dtype=float)

    i = 0

    while i < n:
        j = i

        while j < n and sorted_x[j] == sorted_x[i]:
            j += 1

        midranks[i:j] = 0.5 * (i + j - 1) + 1
        i = j

    result = np.empty(n, dtype=float)
    result[order] = midranks

    return result


def delong_test(y_true, score_a, score_b):
    y_true = np.asarray(y_true)
    scores = np.vstack([
        np.asarray(score_a),
        np.asarray(score_b)
    ])

    positive = y_true == 1
    negative = y_true == 0

    m = positive.sum()
    n = negative.sum()

    if m < 2 or n < 2:
        return np.nan, np.nan, np.nan, np.nan

    sorted_indices = np.argsort(-y_true)
    scores = scores[:, sorted_indices]

    tx = np.array([
        compute_midrank(scores[r, :m])
        for r in range(2)
    ])

    ty = np.array([
        compute_midrank(scores[r, m:])
        for r in range(2)
    ])

    tz = np.array([
        compute_midrank(scores[r, :])
        for r in range(2)
    ])

    aucs = (
        tz[:, :m].sum(axis=1) / (m * n)
        - (m + 1) / (2 * n)
    )

    v01 = (tz[:, :m] - tx) / n
    v10 = 1.0 - (tz[:, m:] - ty) / m

    covariance = (
        np.cov(v01) / m
        + np.cov(v10) / n
    )

    contrast = np.array([1.0, -1.0])

    variance = float(
        contrast @ covariance @ contrast.T
    )

    difference = float(aucs[0] - aucs[1])

    if variance <= 0:
        return float(aucs[0]), float(aucs[1]), difference, np.nan

    z = difference / np.sqrt(variance)
    p_value = 2 * norm.sf(abs(z))

    return float(aucs[0]), float(aucs[1]), difference, float(p_value)


# ---------------------------------------------------------
# 5. McNemar tests
# ---------------------------------------------------------

mcnemar_rows = []

for model_a, model_b in combinations(models, 2):
    a = aligned[f"{model_a}__pred"].to_numpy()
    b = aligned[f"{model_b}__pred"].to_numpy()
    y = aligned["y_true"].to_numpy()

    correct_a = a == y
    correct_b = b == y

    # A correct / B incorrect versus A incorrect / B correct
    discordant_a = int(np.sum(correct_a & ~correct_b))
    discordant_b = int(np.sum(~correct_a & correct_b))

    discordant_total = discordant_a + discordant_b

    p_value = (
        binomtest(
            discordant_a,
            discordant_total,
            p=0.5,
            alternative="two-sided"
        ).pvalue
        if discordant_total > 0 else 1.0
    )

    # Matched-pair odds ratio; Haldane correction for zero cells
    matched_or = (
        (discordant_a + 0.5) / (discordant_b + 0.5)
    )

    accuracy_difference = (
        accuracy_score(y, a) - accuracy_score(y, b)
    )

    mcnemar_rows.append({
        "Model_A": model_a,
        "Model_B": model_b,
        "A_Correct_B_Wrong": discordant_a,
        "A_Wrong_B_Correct": discordant_b,
        "Discordant_Pairs": discordant_total,
        "Accuracy_Difference_A_minus_B": accuracy_difference,
        "Matched_Odds_Ratio_A_vs_B": matched_or,
        "P_Value": p_value
    })

mcnemar_df = pd.DataFrame(mcnemar_rows)


# ---------------------------------------------------------
# 6. DeLong pairwise comparisons
# ---------------------------------------------------------

delong_rows = []

for model_a, model_b in combinations(models, 2):
    auc_a, auc_b, difference, p_value = delong_test(
        aligned["y_true"].to_numpy(),
        aligned[f"{model_a}__score"].to_numpy(),
        aligned[f"{model_b}__score"].to_numpy()
    )

    delong_rows.append({
        "Model_A": model_a,
        "Model_B": model_b,
        "AUC_A": auc_a,
        "AUC_B": auc_b,
        "AUC_Difference_A_minus_B": difference,
        "P_Value": p_value
    })

delong_df = pd.DataFrame(delong_rows)


# ---------------------------------------------------------
# 7. Site-cluster bootstrap for paired differences
# ---------------------------------------------------------

rng = np.random.default_rng(RANDOM_STATE)
sites = aligned["HeldOutSite"].unique()

bootstrap_rows = []

for model_a, model_b in combinations(models, 2):
    differences = {
        "Accuracy": [],
        "BalancedAccuracy": [],
        "ROC_AUC": []
    }

    site_groups = {
        site: aligned[aligned["HeldOutSite"] == site]
        for site in sites
    }

    for _ in range(N_BOOTSTRAP):
        sampled_sites = rng.choice(
            sites, size=len(sites), replace=True
        )

        sample = pd.concat(
            [site_groups[site] for site in sampled_sites],
            ignore_index=True
        )

        y = sample["y_true"].to_numpy()

        a_pred = sample[f"{model_a}__pred"].to_numpy()
        b_pred = sample[f"{model_b}__pred"].to_numpy()

        a_score = sample[f"{model_a}__score"].to_numpy()
        b_score = sample[f"{model_b}__score"].to_numpy()

        differences["Accuracy"].append(
            accuracy_score(y, a_pred) -
            accuracy_score(y, b_pred)
        )

        differences["BalancedAccuracy"].append(
            balanced_accuracy_score(y, a_pred) -
            balanced_accuracy_score(y, b_pred)
        )

        differences["ROC_AUC"].append(
            roc_auc_score(y, a_score) -
            roc_auc_score(y, b_score)
        )

    for metric, values in differences.items():
        values = np.asarray(values)

        bootstrap_rows.append({
            "Model_A": model_a,
            "Model_B": model_b,
            "Metric": metric,
            "Mean_Difference": np.mean(values),
            "CI_2.5": np.quantile(values, 0.025),
            "CI_97.5": np.quantile(values, 0.975),
            "Bootstrap_Replicates": len(values)
        })

bootstrap_df = pd.DataFrame(bootstrap_rows)


# ---------------------------------------------------------
# 8. Multiple-testing correction
# ---------------------------------------------------------

def correct_p_values(df):
    df = df.copy()

    valid = df["P_Value"].notna()

    df["P_Holm"] = np.nan
    df["P_Bonferroni"] = np.nan

    if valid.any():
        p = df.loc[valid, "P_Value"].to_numpy()

        df.loc[valid, "P_Holm"] = multipletests(
            p, method="holm"
        )[1]

        df.loc[valid, "P_Bonferroni"] = multipletests(
            p, method="bonferroni"
        )[1]

    return df


mcnemar_df = correct_p_values(mcnemar_df)
delong_df = correct_p_values(delong_df)


# ---------------------------------------------------------
# 9. Exploratory quantile regression
# ---------------------------------------------------------

quantile_rows = []

for model in models:
    data = (
        site_metrics[site_metrics["Model"] == model]
        .dropna(subset=[
            "BalancedAccuracy", "N_Test", "ASD_Prevalence"
        ])
        .copy()
    )

    data["Log_N_Test"] = np.log(data["N_Test"])

    X = sm.add_constant(
        data[["Log_N_Test", "ASD_Prevalence"]]
    )

    y = data["BalancedAccuracy"]

    if len(data) < 10:
        warnings.warn(
            f"Only {len(data)} sites for {model}; "
            "quantile regression skipped."
        )
        continue

    for q in QUANTILES:
        try:
            result = sm.QuantReg(y, X).fit(
                q=q,
                max_iter=10000
            )

            for variable in result.params.index:
                quantile_rows.append({
                    "Model": model,
                    "Quantile": q,
                    "Variable": variable,
                    "Coefficient": result.params[variable],
                    "Std_Error": result.bse[variable],
                    "P_Value": result.pvalues[variable],
                    "N_Sites": len(data)
                })

        except Exception as error:
            warnings.warn(
                f"Quantile regression failed for {model}, "
                f"q={q}: {error}"
            )

quantile_df = pd.DataFrame(quantile_rows)

if not quantile_df.empty:
    quantile_df = correct_p_values(quantile_df)


# ---------------------------------------------------------
# 10. Save results
# ---------------------------------------------------------

outputs = {
    "10_mcnemar_pairwise.csv": mcnemar_df,
    "10_delong_pairwise.csv": delong_df,
    "10_site_cluster_bootstrap.csv": bootstrap_df,
    "10_quantile_regression.csv": quantile_df,
    "10_recomputed_loso_site_metrics.csv": site_metrics
}

for filename, dataframe in outputs.items():
    dataframe.to_csv(
        RESULTS / filename,
        index=False
    )

print("\n" + "=" * 70)
print("MCNEMAR TESTS")
print("=" * 70)
print(mcnemar_df.to_string(index=False))

print("\n" + "=" * 70)
print("DELONG TESTS")
print("=" * 70)
print(delong_df.to_string(index=False))

print("\n" + "=" * 70)
print("SITE-CLUSTER BOOTSTRAP")
print("=" * 70)
print(bootstrap_df.to_string(index=False))

print("\n" + "=" * 70)
print("QUANTILE REGRESSION")
print("=" * 70)
print(quantile_df.to_string(index=False))

print(f"\nAll results saved to: {RESULTS}")
```
